- Given watchlist (list of tickers), show daily stocks prices for last 5 years
- Columns: Date, Open, High, Low, Close, Volume, Ticker
- Scrape watchlist from Google Finance?

```python
watchlist_cad = [
    'bns', 'cls', 'xgd', 'svr', 'mfc', 'sppp', 'slf', 'tsat', 
    'tvk', 'tfii', 'wsp'
]

watchlist_cad = [x + ".to" for x in watchlist_cad]

watchlist_usd = [
    'baba', 'googl', 'aapl', 'azo', 'bah', 'csco', 'crwv', 
    'dell', 'docs', 'itot', 'mod', 'orcl', 'pltr', 'voo',
    'wm', 'wsm'
]
```

```python
start_dt = date.today() - relativedelta(years=5)
print(start_dt)
```

```python
finance_df = yf.download(watchlist_cad, start=start_dt)
```
