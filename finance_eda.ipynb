```python
def get_database_uri():
    """
    Build a PostgreSQL connection URI from environment variables.

    Reads DB_USER, DB_PASSWORD, DB_HOST, DB_PORT, and DB_NAME from the
    environment (typically loaded from a .env file via python-dotenv) and
    returns a SQLAlchemy-compatible connection string.

    Returns:
        str: Connection URI in the format
             postgresql+psycopg2://user:password@host:port/dbname

    Raises:
        KeyError: If any required environment variable is not set.
    """
    
    db_user = os.getenv("DB_USER")
    db_password = os.getenv("DB_PASSWORD")
    db_host = os.getenv("DB_HOST")
    db_port = os.getenv("DB_PORT")
    db_name = os.getenv("DB_NAME")

    db_uri = f"postgresql+psycopg2://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}"

    return db_uri
```

```python
db_uri = get_database_uri()
engine = create_engine(db_uri)
```

```python
insp = inspect(engine)
tables = insp.get_table_names()
tables = [x for x in tables if "finance" in x]
print(tables)
```
